
### Verify parent data
Atlikon's four tables were uploaded into `fmcg.gold`. This counts their rows, with `UNION ALL` stacking the four results into one table.

```sql
SELECT 'dim_customers' AS table_name, COUNT(*) AS row_count FROM fmcg.gold.dim_customers
UNION ALL
SELECT 'dim_gross_price', COUNT(*) FROM fmcg.gold.dim_gross_price
UNION ALL
SELECT 'dim_products', COUNT(*) FROM fmcg.gold.dim_products
UNION ALL
SELECT 'fact_orders', COUNT(*) FROM fmcg.gold.fact_orders;
```
