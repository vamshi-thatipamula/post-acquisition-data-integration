
### Create catalog and schemas
Creates the `fmcg` catalog with three layers: `bronze` (raw), `silver` (cleaned), `gold` (business-ready).

```python
%sql
CREATE CATALOG IF NOT EXISTS fmcg;
USE CATALOG fmcg;
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS fmcg.gold;
CREATE SCHEMA IF NOT EXISTS fmcg.silver;
CREATE SCHEMA IF NOT EXISTS fmcg.bronze;
```


### Step 2: Verify the setup
List the schemas in the catalog to confirm all three layers were created.

```python
%sql
SHOW SCHEMAS IN fmcg;
```


### Notes
